## 1. Импорт библиотек, исходные данные

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

#-------------------------------------------
pd.set_option('display.max_rows', None)      #чтобы было видно все строки в pandas
pd.set_option('display.max_columns', None)
# Десятичный вывод с 4 знаками после точки; точность вычислений сохраняется.
pd.set_option('display.float_format', '{:.4f}'.format)
#-------------------------------------------

# Мощность двигателя
power_hp = 150.0
HP_TO_KW = 0.73549875
power_kw = power_hp * HP_TO_KW

# Наружные условия
T_a = 288.0          # K
p_a = 101.3          # кПа
velocity = 0.0       # м/с

# Температура газа перед турбиной
T_g = 1250.0         # K

# КПД и потери
eta_k_poly = 0.86    # п кпд цбк по методичке 0.85..0.87
eta_gg_poly = 0.90   # п кпд газогенератора
eta_pt_poly = 0.90   # п кпд силовой турбины

sigma_in = 0.985
sigma_burn = 0.965
eta_burn = 0.98

eta_m_gg = 0.99
eta_m_pt = 0.99

# КПД редуктора:
eta_red = 0.98

# Отборы воздуха
g_leak = 0.005
g_cool = 0.0

# Топливо и газовые постоянные
Q_lhv = 43000.0      # кДж/кг
l0 = 14.61
T_ref = 288.0        # K
R_air = 0.287        # кДж/(кг К)
R_gas = 0.288        # кДж/(кг К)

# Сопло и выход из свободной турбины
pi_nozzle_static = 1.04
c_pt_exit = 250.0    # м/с
phi_nozzle = 0.95
beta_prop = 15.0     # Н/кВт
eta_prop = 0.82

# Настройки вариантного расчёта
n_variants = 801

KCAL_TO_KJ = 4.187

print(f"Мощность на валу: {power_hp:.0f} л.с. = {power_kw:.3f} кВт")
```

## 2. Теплофизические свойства

Формулы теплоёмкостей взяты из приложения 6 методички. Для воздуха используется предел alpha → infty

**Формулы следующей ячейки: теплофизические свойства и деление пополам**

Обозначения: $T$ — К, $\alpha$ — коэффициент избытка воздуха, $a=1/\alpha$; для воздуха $a=0$. Теплоёмкости — **кДж/(кг·К)**. В коде `cp_i` — истинная, `cp_m` — средняя теплоёмкость.

Вспомогательные полиномы температуры:

$$U=2.5\cdot10^{-3}T-0.875,\qquad V=2.5\cdot10^{-5}T^2-0.0275T+6.5625,$$
$$W=1.176\cdot10^{-3}T-0.88235,\qquad Z=5.5556\cdot10^{-6}T^2-0.013056T+6.67.$$

$$c_{pi}(T,\alpha)=4.187\begin{cases}
0.0174a+0.2407+(0.0093a+0.0193)U+\left(0.002-\dfrac{0.001056a}{1-0.2a}\right)V,&T<750,\\[2mm]
0.0267a+0.26+(0.0133a+0.032)W-\left(0.00374+\dfrac{0.0094a^2}{1+10a^2}\right)Z,&T\ge750.
\end{cases}$$

$$
c_{pm}(T,\alpha)=4.187\begin{cases}
\dfrac{2.25a+1.2}{10^5}(T-70)+0.236,&T<700,\\
\dfrac{1.25a+2.2}{10^5}(T+450)+0.218,&T\ge700.
\end{cases}
$$

Функции энтальпии записаны **точно как в существующем коде**:

$$h(T,\alpha)=c_{pm}(T,\alpha)T,\qquad
\Delta h(T,\alpha)=h(T,\alpha)-h(T_{ref},\alpha).$$

`process_properties` использует приближение:

$$T_m=\frac{T_1+T_2}{2},\qquad c_p=c_{pi}(T_m,\alpha),\qquad k=\frac{c_p}{c_p-R}.$$

В `bisect` решается $F(x)=0$: $m=(l+r)/2$; если $F(l)F(m)\le0$, заменяем $r=m$, иначе $l=m$. После 80 шагов возвращается $(l+r)/2$. Это выбранный численный алгоритм, не отдельная физическая формула.

**Источник:** «цикл.pdf», приложение 6, с. 56–57; согласование свойств процесса — §2.3, с. 16. Определения `h`, `dh` и правила осреднения показаны здесь для объяснения уже имеющегося кода, без изменения его модели.


```python
def cp_i(T, alpha=np.inf):
    if np.isinf(alpha):
        a = 0.0
    else:
        a = 1.0 / alpha

    if T < 750.0:
        value = ((0.0174*a + 0.2407)
                 + (0.0093*a + 0.0193)*(2.5e-3*T - 0.875)
                 + (2.0e-3 - 1.056e-3*a/(1.0 - 0.2*a))
                 * (2.5e-5*T*T - 0.0275*T + 6.5625))
    else:
        value = ((0.0267*a + 0.26)
                 + (0.0133*a + 0.032)*(1.176e-3*T - 0.88235)
                 - (0.374e-2 + 0.94e-2*a*a/(1.0 + 10*a*a))
                 * (5.5556e-6*T*T - 1.3056e-2*T + 6.67))

    return KCAL_TO_KJ * value


def cp_m(T, alpha=np.inf):
    a = 0.0 if np.isinf(alpha) else 1.0 / alpha

    if T < 700.0:
        value = (2.25*a + 1.2)/1e5*(T - 70.0) + 0.236
    else:
        value = (1.25*a + 2.2)/1e5*(T + 450.0) + 0.218

    return KCAL_TO_KJ * value


def h(T, alpha=np.inf):
    return cp_m(T, alpha) * T


def dh(T, alpha):
    return h(T, alpha) - h(T_ref, alpha)


def process_properties(T1, T2, alpha, R):
    T_mean = (T1 + T2) / 2
    cp = cp_i(T_mean, alpha)
    k = cp / (cp - R)
    return cp, k


def bisect(fun, left, right, n=80):
    f_left = fun(left)

    for _ in range(n):
        middle = (left + right) / 2
        f_middle = fun(middle)

        if f_left * f_middle <= 0:
            right = middle
        else:
            left = middle
            f_left = f_middle

    return (left + right) / 2
```

## 3. Входное устройство

**Формулы следующей функции `inlet()`**

$T_a,p_a,V$ — статические параметры атмосферы и скорость полёта. Индекс 2 — вход компрессора; возвращаемые $T_2,p_2$ — **полные** параметры.

$$c_p=c_{pi}\left(\frac{T_a+T_2}{2},\infty\right),\qquad k=\frac{c_p}{c_p-R_a},$$
$$c_p(T_2-T_a)=\frac{V^2}{2000},\qquad
M=\frac{V}{\sqrt{1000kR_aT_a}},$$
$$\pi_V=\left(\frac{T_2}{T_a}\right)^{k/(k-1)},\qquad
p_2=p_a\pi_V\sigma_{in}.$$

$T_2$ находится делением пополам. Числа 1000 и 2000 связаны с использованием кДж. При $V=0$: $T_2=T_a$, $\pi_V=1$.

**Источник:** «цикл.pdf», §2.1, с. 10 — полные параметры на входе и коэффициент сохранения давления. Выражения через $V$ — раскрытие перехода от статических к полным параметрам, уже реализованное в исходном ноутбуке.


```python
def inlet():
    def equation(T2):
        cp, k = process_properties(T_a, T2, np.inf, R_air)
        return T2 - T_a - velocity**2/(2000*cp)

    T2 = bisect(equation, T_a, T_a + velocity**2/1000 + 1)
    cp, k = process_properties(T_a, T2, np.inf, R_air)

    M = velocity / math.sqrt(1000*k*R_air*T_a)
    pi_v = (T2/T_a)**(k/(k-1))
    p2 = p_a * pi_v * sigma_in

    return {
        'T2': T2,
        'p2': p2,
        'pi_v': pi_v,
        'M_flight': M,
        'cp_in': cp,
        'k_in': k
    }
```

## 4. Компрессор

**Формулы следующей функции `compressor()`**

$\pi_k=p_3/p_2$ — степень повышения полного давления; $\eta_{kp}$ — политропический КПД. Свойства определяются по средней температуре $T_m=(T_2+T_3)/2$.

$$c_p=c_{pi}(T_m,\infty),\quad k=\frac{c_p}{c_p-R_a},\quad
T_3=T_2\pi_k^{(k-1)/(k\eta_{kp})},$$
$$\eta_k=\frac{\pi_k^{(k-1)/k}-1}{\pi_k^{(k-1)/(k\eta_{kp})}-1},\qquad
p_3=p_2\pi_k,\qquad L_k=c_p(T_3-T_2).$$

$T_3$ решается неявно, поскольку $c_p$ и $k$ зависят от неё. В коде `expm1(x)` означает $e^x-1$ и сохраняет точность при малом $x$.

**Источник:** «цикл.pdf», с. 9 — переход от политропического КПД к адиабатному; §2.2, с. 11–12 — параметры и работа компрессора. Формула $T_3$ — преобразование этих соотношений.


```python
def compressor(pi_k, inp):
    T2 = inp['T2']

    def equation(T3):
        cp, k = process_properties(T2, T3, np.inf, R_air)
        return T3 - T2*pi_k**((k-1)/(k*eta_k_poly))

    T3 = bisect(equation, T2, T_g)
    cp, k = process_properties(T2, T3, np.inf, R_air)

    x = (k-1)/k * math.log(pi_k)
    eta_k = math.expm1(x) / math.expm1(x/eta_k_poly)

    p3 = inp['p2'] * pi_k
    L_k = cp * (T3 - T2)

    return {
        'T3': T3,
        'p3': p3,
        'L_k': L_k,
        'cp_k': cp,
        'k_k': k,
        'eta_k': eta_k
    }
```

## 5. Камера сгорания

**Формулы следующей функции `combustor()`**

`gm` — расход топлива, отнесённый к воздуху, поступающему в камеру сгорания. $h$ и $\Delta h$ — функции предыдущего раздела. В числителе существующего кода используется $h(T,\infty)$:

$$D_Q=Q_H\eta_b-(1+l_0)\Delta h(T_g,1)+l_0\Delta h(T_g,\infty),$$
$$g_m=\frac{h(T_g,\infty)-h(T_3,\infty)}{D_Q},\qquad
\alpha=\frac{1}{g_ml_0}.$$

Доли расходов относительно воздуха через компрессор:

$$b=1-g_{ут}-g_{охл},\qquad
f=bg_m,\qquad g_g=b(1+g_m),\qquad p_g=p_3\sigma_b.$$

Соответствие коду: `fuel_air` $=f$, `gas_air` $=g_g$, `pg` $=p_g$.

**Источник:** «цикл.pdf», §3.1, с. 19–21 — уточнённый расход топлива и состав продуктов сгорания; §3.2, с. 22 — относительные расходы. Формулы записаны в форме реализованной функции, без изменения исходного расчёта.


```python
def combustor(comp):
    denominator = (Q_lhv*eta_burn
                   - (1+l0)*dh(T_g, 1)
                   + l0*dh(T_g, np.inf))

    gm = (h(T_g) - h(comp['T3'])) / denominator
    alpha = 1 / (gm*l0)

    b = 1 - g_leak - g_cool
    fuel_air = b * gm
    gas_air = b * (1 + gm)
    pg = comp['p3'] * sigma_burn

    return {
        'Tg': T_g,
        'pg': pg,
        'gm': gm,
        'alpha': alpha,
        'b': b,
        'fuel_air': fuel_air,
        'gas_air': gas_air
    }
```

## 6. Турбина компрессора

**Формулы следующих функций `turbine_eta()` и `gas_generator_turbine()`**

Для произвольной турбины при $\pi_t=p_{in}^*/p_{out}^*$ и политропическом КПД $\eta_{tp}$:

$$\eta_t^*=\frac{1-\pi_t^{-(k-1)\eta_{tp}/k}}{1-\pi_t^{-(k-1)/k}}.$$

Для турбины компрессора (ТК, газогенератора):

$$L_{ГГ}=\frac{L_k}{g_g\eta_{m,ГГ}},\qquad
c_p=c_{pi}\left(\frac{T_g+T_4}{2},\alpha\right),\qquad k=\frac{c_p}{c_p-R_g},$$
$$c_p(T_g-T_4)=L_{ГГ},\qquad
\pi_{ГГ}=\left(\frac{T_g}{T_4}\right)^{k/[(k-1)\eta_{p,ГГ}]},\qquad
p_4=\frac{p_g}{\pi_{ГГ}}.$$

Температуры и давления здесь полные. $T_4$ находится из баланса работы; затем рассчитывается адиабатный КПД функцией `turbine_eta`.

**Источник:** «цикл.pdf», с. 9 — пересчёт КПД; с. 17–19 — баланс мощностей и параметры компрессорной турбины. Формула $\pi_{ГГ}$ — преобразование политропического соотношения.


```python
def turbine_eta(pi_t, k, eta_poly):
    x = -(k-1)/k * math.log(pi_t)
    return math.expm1(x*eta_poly) / math.expm1(x)


def gas_generator_turbine(comp, burn):
    L_gg = comp['L_k'] / (burn['gas_air'] * eta_m_gg)

    def equation(T4):
        cp, k = process_properties(T_g, T4, burn['alpha'], R_gas)
        return cp*(T_g - T4) - L_gg

    T4 = bisect(equation, T_a, T_g)
    cp, k = process_properties(T_g, T4, burn['alpha'], R_gas)

    pi_gg = (T_g/T4)**(k/((k-1)*eta_gg_poly))
    p4 = burn['pg'] / pi_gg
    eta_gg = turbine_eta(pi_gg, k, eta_gg_poly)

    return {
        'T4': T4,
        'p4': p4,
        'L_gg': L_gg,
        'pi_gg': pi_gg,
        'cp_gg': cp,
        'k_gg': k,
        'eta_gg': eta_gg
    }
```

## 7. Свободная силовая турбина

**Формулы следующих функций `static_from_total()` и `power_turbine()`**

Переход от полных параметров к статическим при скорости $c$:

$$c_p=c_{pi}\left(\frac{T^*+T}{2},\alpha\right),\quad k=\frac{c_p}{c_p-R_g},\quad
c_p(T^*-T)=\frac{c^2}{2000},$$
$$M=\frac{c}{\sqrt{1000kR_gT}},\qquad r=\frac{p^*}{p}=\left(\frac{T^*}{T}\right)^{k/(k-1)}.$$

Для силовой турбины в модели цикла задаются статическое давление и выходная скорость:

$$p_5=\pi_{с,стат}\,p_a,\qquad c_5=c_{pt,exit},\qquad p_5^*=p_5r_5.$$

Неизвестная $T_5^*$ определяется совместно с переходом к статическим параметрам:

$$T_5^*=T_4\left(\frac{p_5^*}{p_4}\right)^{(k-1)\eta_{p,СТ}/k},\qquad
c_p=c_{pi}\left(\frac{T_4+T_5^*}{2},\alpha\right),\quad k=\frac{c_p}{c_p-R_g}.$$

После решения:

$$\pi_{СТ}=\frac{p_4}{p_5^*},\quad L_{СТ}=c_p(T_4-T_5^*),\quad
\pi_{с}=\frac{p_5^*}{p_a},\quad
\eta_{СТ}^*=\frac{1-\pi_{СТ}^{-(k-1)\eta_{p,СТ}/k}}{1-\pi_{СТ}^{-(k-1)/k}}.$$

**Источник:** «цикл.pdf», с. 9 и §2.3, с. 16–19. Полное давление $p_5^*$ и статическое $p_5$ — разные величины; связь раскрыта по энергетическому и изоэнтропическому соотношениям, используемым исходным кодом. Значения $\pi_{с,стат}$ и $c_5$ — входные решения проекта.


```python
def static_from_total(T_total, c, alpha, R):
    def equation(T_static):
        cp, k = process_properties(T_total, T_static, alpha, R)
        return cp*(T_total - T_static) - c*c/2000

    T_static = bisect(equation, 100.0, T_total)
    cp, k = process_properties(T_total, T_static, alpha, R)

    M = c / math.sqrt(1000*k*R*T_static)
    pressure_ratio = (T_total/T_static)**(k/(k-1))

    return T_static, cp, k, M, pressure_ratio


def power_turbine(gg, burn):
    p5 = pi_nozzle_static * p_a

    def equation(T5_total):
        T5, cp5, k5, M5, ratio5 = static_from_total(
            T5_total, c_pt_exit, burn['alpha'], R_gas
        )
        p5_total = p5 * ratio5
        cp, k = process_properties(gg['T4'], T5_total, burn['alpha'], R_gas)

        return T5_total - gg['T4']*(p5_total/gg['p4'])**((k-1)*eta_pt_poly/k)

    T5_total = bisect(equation, max(200.0, T_a), gg['T4'])

    T5, cp5, k5, M5, ratio5 = static_from_total(
        T5_total, c_pt_exit, burn['alpha'], R_gas
    )
    p5_total = p5 * ratio5

    cp, k = process_properties(gg['T4'], T5_total, burn['alpha'], R_gas)
    pi_pt = gg['p4'] / p5_total
    L_pt = cp * (gg['T4'] - T5_total)
    eta_pt = turbine_eta(pi_pt, k, eta_pt_poly)

    return {
        'T5_total': T5_total,
        'p5_total': p5_total,
        'T5': T5,
        'p5': p5,
        'M5': M5,
        'cp5': cp5,
        'k5': k5,
        'pi_nozzle': p5_total/p_a,
        'L_pt': L_pt,
        'pi_pt': pi_pt,
        'cp_pt': cp,
        'k_pt': k,
        'eta_pt': eta_pt
    }
```

## 8. Реактивное сопло

**Формулы следующей функции `nozzle()`**

Для существующей модели полного расширения до $p_a$:

$$\pi_с=\frac{p_5^*}{p_a},\qquad
H_с=c_pT_5^*\left(1-\pi_с^{-(k-1)/k}\right),$$
$$T_с=T_5^*-\frac{\varphi_с^2H_с}{c_p},\qquad
c_с=\varphi_с\sqrt{2000H_с},\qquad p_с=p_a.$$

Свойства уточняются по $(T_5^*+T_с)/2$. Удельная тяга сопла, отнесённая к расходу воздуха:

$$R_{уд}=g_g(c_с-V).$$

**Источник:** «цикл.pdf», §3.2, с. 24–27 — реактивное сопло и удельная тяга. Это описание действующей функции: она предполагает полное расширение и не содержит отдельной ветви запертого сопла и члена $(p_с-p_a)A_с$. Код в рамках добавления формул не изменяется.


```python
def nozzle(pt, burn):
    ratio = pt['pi_nozzle']

    def equation(Tc):
        cp, k = process_properties(pt['T5_total'], Tc, burn['alpha'], R_gas)
        H = cp*pt['T5_total']*(1-ratio**(-(k-1)/k))
        return Tc - pt['T5_total'] + phi_nozzle**2*H/cp

    Tc = bisect(equation, 100.0, pt['T5_total'])
    cp, k = process_properties(pt['T5_total'], Tc, burn['alpha'], R_gas)

    H = cp*pt['T5_total']*(1-ratio**(-(k-1)/k))
    c_nozzle = phi_nozzle * math.sqrt(2000*H)
    R_specific = burn['gas_air'] * (c_nozzle - velocity)

    return {
        'Tc': Tc,
        'pc': p_a,
        'H_nozzle': H,
        'c_nozzle': c_nozzle,
        'cp_nozzle': cp,
        'k_nozzle': k,
        'R_specific': R_specific
    }
```

## 9. Мощность и расходы

**Формулы следующей функции `performance()`**

Работы — кДж/кг, мощности — кВт; $G$ — кг/с.

$$N_{уд}=g_gL_{СТ}\eta_{m,СТ}\eta_r,\qquad
G_в=\frac{N_e}{N_{уд}},\qquad G_{в,г}=bG_в,$$
$$G_f=g_mG_{в,г},\qquad G_g=g_gG_в,\qquad G_{f,ч}=3600G_f,\qquad R_с=G_вR_{уд}.$$

Добавка эквивалентной мощности в существующем коде:

$$\Delta N_{уд}=\begin{cases}
R_{уд}/\beta_{винт},&V=0,\\
R_{уд}V/(1000\eta_{винт}),&V\ne0,
\end{cases}$$
$$N_{экв,уд}=N_{уд}+\Delta N_{уд},\qquad N_{экв}=G_вN_{экв,уд}.$$

$$C_{e,вал}=\frac{3600G_f}{N_e},\qquad
C_{e,экв}=\frac{3600G_f}{N_{экв}},\qquad
C_{e,лс}=0.73549875\,C_{e,вал},$$
$$\eta_{вал}=\frac{3600}{C_{e,вал}Q_H},\qquad
\eta_{экв}=\frac{3600}{C_{e,экв}Q_H}.$$

Здесь $C_e$ — кг/(кВт·ч), $Q_H$ — кДж/кг. Мощности агрегатов до механических потерь:

$$N_k=G_вL_k,\qquad N_{ГГ}=G_gL_{ГГ},\qquad N_{СТ}=G_gL_{СТ}.$$

**Источник:** «цикл.pdf», §3.2, с. 22–26 — удельные мощности и эквивалентная мощность ТВД; §3.3, с. 42–44 — экономичность; §4, с. 45 — расходы. Пересчёты на вал, редуктор и единицы — раскрытие используемых балансов. Значения $\beta_{винт}$ и $\eta_{винт}$ заданы в исходном ноутбуке, здесь они не объявляются результатом расчёта винта.


```python
def performance(comp, burn, gg, pt, noz):
    N_specific = burn['gas_air'] * pt['L_pt'] * eta_m_pt * eta_red

    G_air = power_kw / N_specific
    G_comb_air = burn['b'] * G_air
    G_fuel = burn['gm'] * G_comb_air
    G_gas = burn['gas_air'] * G_air

    R_nozzle = noz['R_specific'] * G_air

    if velocity == 0:
        equivalent_add = noz['R_specific'] / beta_prop
    else:
        equivalent_add = noz['R_specific']*velocity/(1000*eta_prop)

    N_equiv_specific = N_specific + equivalent_add
    N_equiv = G_air * N_equiv_specific

    Ce_shaft = 3600*G_fuel/power_kw
    Ce_equiv = 3600*G_fuel/N_equiv

    return {
        'N_specific': N_specific,
        'N_equiv_specific': N_equiv_specific,
        'N_shaft': power_kw,
        'N_equiv': N_equiv,
        'N_equiv_hp': N_equiv/HP_TO_KW,
        'G_air': G_air,
        'G_comb_air': G_comb_air,
        'G_fuel': G_fuel,
        'G_gas': G_gas,
        'fuel_hour': 3600*G_fuel,
        'R_nozzle': R_nozzle,
        'Ce_shaft': Ce_shaft,
        'Ce_equiv': Ce_equiv,
        'Ce_hp': Ce_shaft*HP_TO_KW,
        'eta_shaft': 3600/(Ce_shaft*Q_lhv),
        'eta_equiv': 3600/(Ce_equiv*Q_lhv),
        'N_compressor': G_air*comp['L_k'],
        'N_gg': G_gas*gg['L_gg'],
        'N_pt': G_gas*pt['L_pt']
    }
```

## 10. Полный расчёт одного варианта

**Порядок следующей функции `cycle(pi_k)`**


$$\pi_k\longrightarrow\text{вход}\longrightarrow\text{компрессор}\longrightarrow
\text{камера сгорания}\longrightarrow\text{ТК}\longrightarrow\text{СТ}\longrightarrow
\text{сопло}\longrightarrow\text{мощность и расходы}.$$


```python
def cycle(pi_k):
    inp = inlet()
    comp = compressor(pi_k, inp)
    burn = combustor(comp)
    gg = gas_generator_turbine(comp, burn)
    pt = power_turbine(gg, burn)
    noz = nozzle(pt, burn)
    perf = performance(comp, burn, gg, pt, noz)

    result = {'pi_k': pi_k}
    result.update(inp)
    result.update(comp)
    result.update(burn)
    result.update(gg)
    result.update(pt)
    result.update(noz)
    result.update(perf)

    return result
```

## 11. Вариантный расчёт и выбор pi_k

Сначала считаем диапазон значений pi_k, затем ищем минимум расхода воздуха и минимум удельного расхода топлива.

```python
T2 = inlet()['T2']
upsilon = T_g / T2
pi_low = upsilon
pi_high = upsilon**2.4

pi_grid = np.linspace(pi_low, pi_high, n_variants)
records = []

for pi_k in pi_grid:
    records.append(cycle(pi_k))

variants = pd.DataFrame(records)

idx_L = variants['G_air'].idxmin()
idx_eta = variants['Ce_shaft'].idxmin()

opt_L = variants.loc[idx_L]
opt_eta = variants.loc[idx_eta]

opt_L = variants.loc[idx_L]
opt_eta = variants.loc[idx_eta]

# После вариантного расчёта выбираем единую точку для всех результатов
pi_k_selected = 6.0
selected_pi = pi_k_selected

s = cycle(selected_pi)

print(f"Диапазон πк*: {pi_low:.3f} ... {pi_high:.3f}")
print(f"Минимум расхода воздуха: πк* = {opt_L['pi_k']:.3f}")
print(f"Минимум удельного расхода топлива: πк* = {opt_eta['pi_k']:.3f}")
print(f"Принятое значение: πк* = {selected_pi:.3f}")
```

# Проектирование турбины

Предварительный расчёт по заданной удельной работе для ТК и СТ при выбранной точке цикла. Принятый закон проточной части $D_{вт}=\mathrm{const}$ применяется при построении меридионального сечения на следующем этапе.

[Методичка](https://chatgpt.com/library?file_id=libfile_f536fb3e60c881918ed5fdff9ab72d0f&account_id=personal) — Михальцев В. Е., Моляков В. Д. «Теория и проектирование газовой турбины», ч. 1, МГТУ им. Н. Э. Баумана, 2006. Номера страниц ниже приведены по изданию.

## 1. Исходные данные из цикла

ТК — турбина компрессора. СТ — свободная силовая турбина. Входные полные параметры СТ на этом этапе берём из цикла за ТК. Состав газа задают $\alpha$ и $R_g$. Расход один для обеих турбин в принятом варианте без дополнительного подвода охлаждающего воздуха между ними.

$$L_{\mathrm{ТК,треб}}=\frac{N_k}{G_g\eta_{m,\mathrm{ТК}}},\qquad
L_{\mathrm{СТ,треб}}=\frac{N_e}{G_g\eta_{m,\mathrm{СТ}}\eta_r}.$$

Удельные работы уже посчитаны в цикле. Проверяем их по балансам валов. Мощность выражена в кВт, удельная работа в кДж/кг, расход в кг/с.

**Источник.** Михальцев, Моляков, «Теория и проектирование газовой турбины», ч. 1, с. 84, § 9, «Задание» и с. 93, п. 152–153.

```python
design_cycle = s
pi_k_design = design_cycle['pi_k']
G_gas_design = design_cycle['G_gas']
alpha_design = design_cycle['alpha']

L_tk_required = design_cycle['L_gg']
L_st_required = design_cycle['L_pt']
assert math.isclose(G_gas_design * L_tk_required * eta_m_gg, design_cycle['N_compressor'])
assert math.isclose(G_gas_design * L_st_required * eta_m_pt * eta_red, power_kw)

stage_inputs = pd.DataFrame([
    ['ТК', design_cycle['Tg'], design_cycle['pg'], L_tk_required],
    ['СТ', design_cycle['T4'], design_cycle['p4'], L_st_required]
], columns=['Турбина', 'T0*, К', 'p0*, кПа', 'L треб., кДж/кг']).set_index('Турбина')
stage_inputs['G газа, кг/с'] = G_gas_design
stage_inputs['Коэффициент избытка воздуха'] = alpha_design
stage_inputs['R газа, кДж/(кг·К)'] = R_gas

print(f'Расчётная точка: πк = {pi_k_design:g}, мощность на валу {power_kw:.3f} кВт')
display(stage_inputs)
```

## 2. Предварительный теплоперепад, выходная площадь и обороты

Выбираем начальные $\eta_{л,оц}$ и $c_{2a,оц}$. Как и в п. 15, в оценке выходной кинетической энергии используем осевую составляющую скорости.

$$H_{0,оц}=\frac{L_{треб}+c_{2a,оц}^2/2000}{\eta_{л,оц}},\qquad
T_{2,оц}=T_0^*-\frac{\eta_{л,оц}H_{0,оц}}{c_p},$$
$$p_{2,оц}=p_0^*\left(1-\frac{H_{0,оц}}{c_p T_0^*}\right)^{k/(k-1)},\qquad
v_{2,оц}=\frac{R_gT_{2,оц}}{p_{2,оц}},\qquad A_{2,оц}=\frac{G_gv_{2,оц}}{c_{2a,оц}},$$
$$n_{max,оц}=\sqrt{\frac{\sigma_{л,доп}}{k_n A_{2,оц}}},\qquad \sigma_{л,оц}=k_n n^2 A_{2,оц}.$$

Оценочные значения $\eta_{л,оц}=0{,}90$, $c_{2a,оц}=200$ м/с для ТК и $250$ м/с для СТ приняты для первого приближения. Обороты округлены до 78 000 и 70 000 об/мин соответственно. $k_n=6{,}8\cdot10^{-6}$ применяем с площадью в м², оборотами в об/мин и напряжением в МПа. $\sigma_{л,доп}=300$ МПа — предварительное допущение до выбора материала и ресурса. Для свойств используем среднюю температуру и принятую в цикле функцию `process_properties`.

**Источник.** Михальцев, Моляков, «Теория и проектирование газовой турбины», ч. 1, с. 55–56, формула (63) и коэффициент $k_n$, с. 85–86, п. 13–25.

```python
eta_blade_est = 0.90
c2a_est = {'ТК': 200.0, 'СТ': 250.0}  # м/с
n_selected = {'ТК': 78000.0, 'СТ': 70000.0}  # об/мин
kn = 6.8e-6
sigma_limit = 300.0  # МПа

preliminary_rows = []
for name, row in stage_inputs.iterrows():
    T0 = row['T0*, К']
    p0 = row['p0*, кПа']
    L_required = row['L треб., кДж/кг']
    ca = c2a_est[name]
    n = n_selected[name]
    H0 = (L_required + ca**2 / 2000) / eta_blade_est

    # Уточнение T2 с теплоёмкостью при средней температуре.
    def temperature_residual(T2):
        cp, _ = process_properties(T0, T2, alpha_design, R_gas)
        return cp * (T0 - T2) - eta_blade_est * H0

    if temperature_residual(100.0) <= 0:
        raise ValueError(f'{name}: не найден диапазон статической температуры')
    T2 = bisect(temperature_residual, 100.0, T0)
    cp, k = process_properties(T0, T2, alpha_design, R_gas)
    pressure_base = 1 - H0 / (cp * T0)
    if pressure_base <= 0:
        raise ValueError(f'{name}: теплоперепад слишком велик для принятых исходных данных')
    p2 = p0 * pressure_base**(k / (k - 1))
    v2 = R_gas * T2 / p2
    A2 = G_gas_design * v2 / ca
    n_max = math.sqrt(sigma_limit / (kn * A2))

    preliminary_rows.append({
        'Турбина': name, 'H0 оценка, кДж/кг': H0, 'cp, кДж/(кг·К)': cp, 'k': k,
        'c2a оценка, м/с': ca, 'T2 оценка, К': T2, 'p2 оценка, кПа': p2,
        'v2 оценка, м³/кг': v2, 'A2 оценка, см²': A2 * 1e4,
        'n, об/мин': n, 'n_max оценка, об/мин': n_max,
        'Напряжение, МПа': kn * n**2 * A2, 'n ниже n_max': n < n_max
    })

preliminary = pd.DataFrame(preliminary_rows).set_index('Турбина')
display(preliminary)
```

## 3. Реактивность, оптимальная окружная скорость и средний диаметр



Берём четыре приближённые точки с нижней сплошной кривой $(u/c_0)_{опт}$ на рис. 18. Аппроксимируем их полиномом первой степени. Все расчётные значения $\rho$ лежат в интервале оцифрованных точек.

$$c_0=\sqrt{2000H_{0,оц}},\qquad u_{опт}=c_0\left(\frac{u}{c_0}\right)_{опт}(\rho),\qquad
D_{ср,2}=\frac{60u_{опт}}{\pi n},\qquad l_2=\frac{A_{2,оц}}{\pi D_{ср,2}}.$$

На с. 58 даны ориентировочные пары $(\rho,l/D)$. Между ними используем линейную интерполяцию. При выбранных оборотах методом бисекции ищем $\rho$, при котором $l_2/D_{ср,2}$ совпадает с интерполированным ориентиром. Это способ численного согласования для первого приближения. После расчёта треугольников скоростей нужно отдельно проверить реактивность у корня и уточнить параметры ступени.

**Источник.** Михальцев, Моляков, «Теория и проектирование газовой турбины», ч. 1, с. 52–53, рис. 18, с. 58–59, § 6.2 и с. 86, п. 26–32.

```python
# Приближённое считывание нижней кривой рис. 18.
rho_fig18 = np.array([0.0, 0.2, 0.4, 0.6])
u_c0_fig18 = np.array([0.450, 0.525, 0.600, 0.675])
u_c0_fit = np.polyfit(rho_fig18, u_c0_fig18, 1)
print(f'Аппроксимация рис. 18: (u/c0)опт ≈ {u_c0_fit[1]:.3f} + {u_c0_fit[0]:.3f}·ρ')
display(pd.DataFrame({'ρ': rho_fig18, '(u/c0)опт, по рисунку': u_c0_fig18}))

# Таблица ориентиров со с. 58. Степень реактивности отсортирована по возрастанию.
rho_ref = np.array([0.2, 0.3, 0.4, 0.5])
lD_ref = np.array([1/9, 1/6, 1/4, 1/3])

geometry_rows = []
for name, row in preliminary.iterrows():
    if not row['n ниже n_max']:
        raise ValueError(f'{name}: снизьте выбранные обороты до оценки n_max')
    c0 = math.sqrt(2000 * row['H0 оценка, кДж/кг'])
    n = row['n, об/мин']
    A2 = row['A2 оценка, см²'] / 1e4

    def lD_residual(rho):
        u_opt = c0 * np.polyval(u_c0_fit, rho)
        D2 = 60 * u_opt / (math.pi * n)
        return A2 / (math.pi * D2**2) - np.interp(rho, rho_ref, lD_ref)

    if lD_residual(rho_ref[0]) * lD_residual(rho_ref[-1]) > 0:
        raise ValueError(f'{name}: при выбранных оборотах нет пересечения в диапазоне ρ=0,2–0,5')
    rho = bisect(lD_residual, rho_ref[0], rho_ref[-1])
    u_c0_opt = np.polyval(u_c0_fit, rho)
    u_opt = c0 * u_c0_opt
    D2 = 60 * u_opt / (math.pi * n)
    l2 = A2 / (math.pi * D2)
    geometry_rows.append({
        'Турбина': name, 'ρ предварительная': rho, 'c0, м/с': c0,
        '(u/c0) оптимальное': u_c0_opt, 'u оптимальная, м/с': u_opt,
        'D средний за РК, мм': D2 * 1000, 'l рабочей лопатки, мм': l2 * 1000,
        'l/D расчётное': l2 / D2, 'l/D по ориентиру': np.interp(rho, rho_ref, lD_ref)
    })

geometry = pd.DataFrame(geometry_rows).set_index('Турбина')
display(geometry)
```

Предварительные независимые размеры из п. 3 используются для выбора общей втулки в п. 4. После этого геометрия обеих ступеней пересчитывается с одним диаметром.

## 4. Постоянная втулка и меридиональное сечение

Общий внутренний диаметр обеих ступеней и перехода определяем по предварительному выходному сечению СТ из п. 3. Это выбранное замыкающее условие для двухвальной схемы. При общей втулке пересчитываем средние диаметры и высоты по площади, реактивность по ориентиру со с. 58 и оценочные обороты по оптимальной окружной скорости. Значения вне диапазона таблицы требуют нового выбора исходных данных.

$$D_{вт}=D_{ср,2}-l_2,\qquad
A=\frac{\pi}{4}(D_{нар}^2-D_{вт}^2),\qquad
l=\frac{\sqrt{D_{вт}^2+4A/\pi}-D_{вт}}{2},\qquad D_{ср}=D_{вт}+l.$$

Принимаем предварительные удлинения $l/b_a=2{,}25$ для СА и $2{,}75$ для РК, осевой зазор между ними $0{,}225b_{a,СА}$. Меридиональная схема после п. 6 использует уточнённые высоты. Длину межтурбинного перехода 8 мм принимаем для эскиза. Входную высоту СА на схеме принимаем равной выходной.

**Источник.** Михальцев, Моляков, «Теория и проектирование газовой турбины», ч. 1, § 6.3, с. 61–63, § 9, п. 63–69, с. 88.

```python
D_hub_common = (geometry.loc['СТ', 'D средний за РК, мм'] - geometry.loc['СТ', 'l рабочей лопатки, мм']) / 1000
for name, row in preliminary.iterrows():
    A2 = row['A2 оценка, см²'] / 1e4
    l2 = (math.sqrt(D_hub_common**2 + 4*A2/math.pi) - D_hub_common)/2
    D2 = D_hub_common + l2
    ratio = l2/D2
    # На с. 58 приведены ориентиры, а не точное уравнение.
    # Для ТК ниже табличного диапазона принимаем нижний ориентир ρ=0,2,
    # затем проверяем реактивность у корня.
    rho = float(np.interp(ratio, lD_ref, rho_ref))
    c0 = math.sqrt(2000*row['H0 оценка, кДж/кг'])
    u = c0*np.polyval(u_c0_fit,rho)
    n_selected[name] = 60*u/(math.pi*D2)
    geometry.loc[name, ['ρ предварительная','D средний за РК, мм','l рабочей лопатки, мм',
                        'l/D расчётное','l/D по ориентиру','u оптимальная, м/с','(u/c0) оптимальное']] = [
        rho,D2*1000,l2*1000,ratio,np.interp(rho,rho_ref,lD_ref),u,u/c0]
hub_diameter = pd.Series(D_hub_common,index=geometry.index)
display(geometry)
display(pd.DataFrame({'D втулки, мм':hub_diameter*1000,'n уточнённые, об/мин':pd.Series(n_selected)}))
aspect_ratio_sa = 2.25
aspect_ratio_rk = 2.75
axial_gap_ratio = 0.225
transition_length = 0.008  # м, для эскиза
```

## 5. Газодинамический расчёт СА и РК на среднем радиусе

Принимаем углы выхода из СА 18° для ТК и 24° для СТ, $c_{2a}/c_{1a}=1{,}10$, коэффициенты скорости $\varphi=0{,}97$, $\psi=0{,}96$ и абсолютный радиальный зазор 0,5 мм по нижнему ориентиру для малых турбин на с. 63. Это исходные оценки для проверяемого варианта. Реактивность и обороты берём после согласования общей втулки в п. 4. Модель использует один расход через венцы, полную подачу и средние свойства газа.

Индекс 1 относится к выходу СА, индекс 2 — к выходу РК. Осевая компонента положительна по потоку, окружная — по вращению. Углы потока отсчитываются от положительного окружного направления.

$$H_c=H_0(1-\rho),\quad c_1=\varphi\sqrt{2000H_c},\quad
T_1=T_0^*-\varphi^2H_c/c_p,\quad
T_{1ад}=T_0^*-H_c/c_p,\quad
p_1=p_0^*(T_{1ад}/T_0^*)^{k/(k-1)}.$$
$$c_{1a}=c_1\sin\alpha_1,\quad c_{1u}=c_1\cos\alpha_1,\quad
A_1=G_g R_g T_1/(p_1c_{1a}),\quad u_i=\pi D_i n/60.$$
$$w_1^2=c_{1a}^2+(c_{1u}-u_1)^2,\quad
H_л=\rho H_0\frac{T_1}{T_{1ад}},\quad
p_2=p_1\left(1-\frac{H_л}{c_pT_1}\right)^{k/(k-1)}.$$
$$w_2^2=\psi^2(w_1^2+2000H_л+u_2^2-u_1^2),\quad
T_2=T_1+\frac{u_2^2-u_1^2-w_2^2+w_1^2}{2000c_p},\quad
A_2=\frac{G_g R_g T_2}{p_2c_{2a}}.$$

Уточняем $l_2$ по площади $A_2$, так как $T_2$ зависит от окружной скорости на выходе. По треугольнику получаем $c_{2u}=u_2-\sqrt{w_2^2-c_{2a}^2}$. Работа по Эйлеру с учётом принятого знака $c_{2u}$ равна $L_u=(u_1c_{1u}-u_2c_{2u})/1000$.

$$L=L_u-h_з-h_{тв},\quad
T_2^*=T_2+\frac{h_з+h_{тв}+c_2^2/2000}{c_p},\quad
p_2^*=p_2\left(\frac{T_2^*}{T_2+(h_з+h_{тв})/c_p}\right)^{k/(k-1)}.$$

Теплоперепады и работы в кДж/кг, давления в кПа, размеры в м. Полный КПД считаем по полному выходному давлению, мощностной КПД по $L/H_0$.

**Источник.** Михальцев, Моляков, «Теория и проектирование газовой турбины», ч. 1, § 9, п. 70–129, с. 88–91.

```python
alpha1_selected = {'ТК': 18.0, 'СТ': 24.0}  # градусы от окружного направления
axial_velocity_ratio = 1.10
phi, psi = 0.97, 0.96
radial_clearance = 0.0005  # м, нижний ориентир для малой турбины, с. 63

def calculate_stage(name, H0, T0, p0):
    rho = geometry.loc[name, 'ρ предварительная']
    n = n_selected[name]
    D_hub = hub_diameter.loc[name]
    alpha1 = math.radians(alpha1_selected[name])
    cp = cp_i(T0, alpha_design)
    for _ in range(20):
        T_mean = T0 - H0 * phi**2 / (2 * cp)
        cp = cp_i(T_mean, alpha_design)
    k = cp / (cp - R_gas)

    # Сопловой аппарат
    Hc = H0 * (1 - rho)
    c1 = phi * math.sqrt(2000 * Hc)
    T1 = T0 - phi**2 * Hc / cp
    T1_ad = T0 - Hc / cp
    p1 = p0 * (T1_ad / T0)**(k / (k - 1))
    v1 = R_gas * T1 / p1
    c1a = c1 * math.sin(alpha1)
    c1u = c1 * math.cos(alpha1)
    A1 = G_gas_design * v1 / c1a
    l1 = (math.sqrt(D_hub**2 + 4 * A1 / math.pi) - D_hub) / 2
    D1 = D_hub + l1
    u1 = math.pi * D1 * n / 60
    w1 = math.hypot(c1a, c1u - u1)

    # Рабочее колесо
    Hl = rho * H0 * T1 / T1_ad
    p2 = p1 * (1 - Hl / (cp * T1))**(k / (k - 1))
    c2a = axial_velocity_ratio * c1a
    l2 = geometry.loc[name, 'l рабочей лопатки, мм'] / 1000
    for _ in range(60):
        D2 = D_hub + l2
        u2 = math.pi * D2 * n / 60
        w2 = psi * math.sqrt(w1**2 + 2000 * Hl + u2**2 - u1**2)
        T2 = T1 + (u2**2 - u1**2 - w2**2 + w1**2) / (2000 * cp)
        v2 = R_gas * T2 / p2
        A2 = G_gas_design * v2 / c2a
        l_next = (math.sqrt(D_hub**2 + 4 * A2 / math.pi) - D_hub) / 2
        if abs(l_next - l2) < 1e-12:
            break
        l2 = l_next
    else:
        raise ValueError(f'{name}: площадь за РК не сошлась')
    if c2a >= w2:
        raise ValueError(f'{name}: осевая скорость за РК выше относительной')
    w2u = -math.sqrt(w2**2 - c2a**2)
    c2u = u2 + w2u
    c2 = math.hypot(c2a, c2u)

    # Удельные потери в зазоре и на трение диска при полной подаче
    Lu = (u1 * c1u - u2 * c2u) / 1000
    hc = (1 / phi**2 - 1) * c1**2 / 2000
    hl = (1 / psi**2 - 1) * w2**2 / 2000
    hv = c2**2 / 2000
    hz = 1.37 * (1 + 1.6 * rho) * (1 + l2 / D2) * (radial_clearance / l2) * Lu
    density_mean = (1 / v1 + 1 / v2) / 2
    friction_power = 1.07 * D2**2 * (u2 / 100)**3 * density_mean
    h_friction = friction_power / G_gas_design
    L = Lu - hz - h_friction
    T_exit = T2 + (hz + h_friction) / cp
    T_exit_total = T_exit + hv / cp
    p_exit_total = p2 * (T_exit_total / T_exit)**(k / (k - 1))
    H_total = cp * T0 * (1 - (p_exit_total / p0)**((k - 1) / k))

    return dict(H0=H0, T0=T0, p0=p0, cp=cp, k=k, rho=rho, n=n,
                D_hub=D_hub, D1=D1, D2=D2, l1=l1, l2=l2, A1=A1, A2=A2,
                T1=T1, T2=T2, p1=p1, p2=p2, c1=c1, c1a=c1a, c1u=c1u,
                c2=c2, c2a=c2a, c2u=c2u, w1=w1, w2=w2, u1=u1, u2=u2,
                alpha1=alpha1_selected[name], alpha2=math.degrees(math.atan2(c2a,c2u)),
                beta1=math.degrees(math.atan2(c1a,c1u-u1)),
                beta2=math.degrees(math.atan2(c2a,w2u)),
                Lu=Lu, L=L, hc=hc, hl=hl, hv=hv, hz=hz, h_friction=h_friction,
                T2_total=T_exit_total, p2_total=p_exit_total, eta=L/H_total,
                eta_power=L/H0, sigma=kn*n**2*A2,
                M1=c1/math.sqrt(1000*k*R_gas*T1),
                M2w=w2/math.sqrt(1000*k*R_gas*T2))

first_stage_results = {}
for name, row in stage_inputs.iterrows():
    H0 = preliminary.loc[name, 'H0 оценка, кДж/кг']
    first_stage_results[name] = calculate_stage(name, H0, row['T0*, К'], row['p0*, кПа'])

display(pd.DataFrame([
    [name, g['L'], g['p2_total'], g['c2'], g['M1'], g['M2w']]
    for name, g in first_stage_results.items()
], columns=['Турбина', 'L, кДж/кг', 'p2*, кПа', 'c2, м/с', 'M1', 'Mw2']))
```

## 6. Согласование по требуемой работе и проверка с циклом

Теплоперепад подбираем бисекцией из $L(H_0)=L_{треб}$. Это численная реализация уточнения по требуемой работе. Сначала рассчитываем ТК. На вход СТ передаём рассчитанные полные параметры за ТК. В первом приближении для перехода принимаем коэффициент сохранения полного давления 1,00.

Выходные давления и скорость СТ сравниваем с циклом. Одна переменная $H_0$ задаёт одно условие по работе. Отклонения давления и скорости показывают, насколько требуется дальнейшее согласование геометрии и треугольников с циклом.

**Источник.** Михальцев, Моляков, «Теория и проектирование газовой турбины», ч. 1, § 9, с. 92–93, п. 130–132 и 142–153.

```python
transition_pressure_ratio = 1.0
stages = {}
for name, row in stage_inputs.iterrows():
    H_start = preliminary.loc[name, 'H0 оценка, кДж/кг']
    L_required = row['L треб., кДж/кг']
    T0, p0 = row['T0*, К'], row['p0*, кПа']
    if name == 'СТ':
        T0 = stages['ТК']['T2_total']
        p0 = transition_pressure_ratio * stages['ТК']['p2_total']

    def work_residual(H0):
        return calculate_stage(name, H0, T0, p0)['L'] - L_required

    left, right = 0.7 * H_start, 1.35 * H_start
    if work_residual(left) * work_residual(right) > 0:
        raise ValueError(f'{name}: работа не согласуется в выбранном диапазоне H0')
    H0 = bisect(work_residual, left, right)
    stages[name] = calculate_stage(name, H0, T0, p0)

stage_summary = pd.DataFrame([
    [name, g['H0'], g['L'], stage_inputs.loc[name, 'L треб., кДж/кг'],
     g['eta'] * 100, g['eta_power'] * 100, g['D1']*1000, g['D2']*1000,
     g['l1']*1000, g['l2']*1000, g['l2']/g['D2'], g['sigma'], g['M1'], g['M2w']]
    for name, g in stages.items()
], columns=['Турбина','H0, кДж/кг','L, кДж/кг','L треб., кДж/кг','η полная, %',
            'η мощностная, %','D1 ср., мм','D2 ср., мм','l1, мм','l2, мм','l2/D2','σл, МПа','M1','Mw2'])
display(stage_summary)

N_tk = stages['ТК']['L'] * G_gas_design * eta_m_gg
N_shaft = stages['СТ']['L'] * G_gas_design * eta_m_pt * eta_red
comparison = pd.DataFrame([
    ['Мощность компрессора, кВт', N_tk, design_cycle['N_compressor']],
    ['Мощность после редуктора, кВт', N_shaft, power_kw],
    ['Полное давление за ТК, кПа', stages['ТК']['p2_total'], design_cycle['p4']],
    ['Полное давление за СТ, кПа', stages['СТ']['p2_total'], design_cycle['p5_total']],
    ['Статическое давление за СТ, кПа', stages['СТ']['p2'], design_cycle['p5']],
    ['Абсолютная скорость за СТ, м/с', stages['СТ']['c2'], c_pt_exit],
], columns=['Параметр','Ступени','Цикл'])
comparison['Отклонение, %'] = (comparison['Ступени'] / comparison['Цикл'] - 1) * 100
display(comparison)
print(f'Мощность после редуктора {N_shaft / HP_TO_KW:.2f} л.с.')
p2_st = stages['СТ']['p2']
print(f'За СТ p₂ = {p2_st:.2f} кПа при pₐ = {p_a:.2f} кПа. Разность {p2_st-p_a:+.2f} кПа.')
```

Совпадение по мощности проверяет работу ступеней. Выходное давление и скорость проверяем отдельно. При $p_{2,СТ}<p_a$ требуется увязка с выходным трактом и его граничным условием.

**Геометрия и треугольники после согласования по работе.** По оси $x$ отложены длины в миллиметрах. Переход между ступенями показан прямыми отрезками. Для эскиза сравниваем модуль угла с принятым ориентиром 20°. Угол наружного контура РК оценён по разности радиусов между выходом СА и выходом РК, отнесённой к длине зазора и РК. На треугольниках $\vec c=\vec u+\vec w$.

```python
fig, ax = plt.subplots(figsize=(9, 3.8))
passage_rows = []
x_start = 0.0
previous_exit = None
colors = {'ТК': '#2166a5', 'СТ': '#d08026'}
for name, g in stages.items():
    r_hub = g['D_hub'] / 2
    b_sa = g['l1'] / aspect_ratio_sa
    b_rk = g['l2'] / aspect_ratio_rk
    gap = axial_gap_ratio * b_sa
    x1 = x_start + b_sa
    x_rk_in = x1 + gap
    x2 = x_rk_in + b_rk
    r_outer1 = r_hub + g['l1']
    r_outer2 = r_hub + g['l2']
    x = np.array([x_start, x1, x2]) * 1000
    r_outer = np.array([r_outer1, r_outer1, r_outer2]) * 1000
    ax.fill_between(x, r_hub*1000, r_outer, alpha=0.18, color=colors[name])
    ax.plot(x, r_outer, 'o-', color=colors[name], label=f'{name}, наружный контур')
    ax.plot(x[[0,2]], [r_hub*1000]*2, color=colors[name])
    ax.axvline(x_rk_in*1000, color=colors[name], lw=0.7, ls=':')
    if previous_exit:
        x_prev, rh_prev, rt_prev = previous_exit
        ax.plot([x_prev,x_start*1000],[rh_prev,r_hub*1000], 'k--', lw=0.9)
        ax.plot([x_prev,x_start*1000],[rt_prev,r_outer1*1000], 'k--', lw=0.9)
    for section, x_section, l in [('Выход СА',x1,g['l1']),('Выход РК',x2,g['l2'])]:
        passage_rows.append([name, section, x_section*1000, r_hub*1000, (r_hub+l)*1000])
    g['b_sa'], g['b_rk'] = b_sa, b_rk
    g['flare_angle'] = math.degrees(math.atan2(r_outer2-r_outer1, gap+b_rk))
    previous_exit = (x2*1000,r_hub*1000,r_outer2*1000)
    x_start = x2 + transition_length
ax.axhline(0, color='black', lw=0.8)
ax.set(xlabel='Осевая координата x, мм', ylabel='Радиус, мм', ylim=(0,None))
ax.grid(alpha=0.25)
ax.legend()
plt.show()

display(pd.DataFrame(passage_rows, columns=['Турбина','Сечение','x, мм','r втулки, мм','r наружный, мм']))
display(pd.DataFrame([
    [name,g['flare_angle'],abs(g['flare_angle'])<=20,g['sigma']<=sigma_limit,g['M1']<1,g['M2w']<1]
    for name,g in stages.items()
], columns=['Турбина','Угол наружного контура, град','|Угол| ≤ 20° (эскизный критерий)',
            'σл ≤ принятого допуска','M1 < 1','Mw2 < 1']))

fig, axes = plt.subplots(2, 2, figsize=(10, 7), constrained_layout=True)
for row, (name,g) in enumerate(stages.items()):
    for col, i in enumerate([1,2]):
        ax = axes[row,col]
        a, cu, u = g[f'c{i}a'], g[f'c{i}u'], g[f'u{i}']
        ax.plot([0,a],[0,cu],'-o', label='c', color='#2166a5')
        ax.plot([0,0],[0,u],'-o', label='u', color='#d08026')
        ax.plot([0,a],[u,cu],'-o', label='w', color='#21916f')
        ax.set(title=f'{name}, сечение {i}', xlabel='Осевая скорость, м/с',
               ylabel='Окружная скорость, м/с')
        ax.set_aspect('equal', adjustable='datalim')
        ax.grid(alpha=0.25)
        ax.legend()
plt.show()
```

## 7. Проверка распределения параметров по радиусу

Используем закон $\alpha_1=\mathrm{const}$ для выхода СА. На корне, среднем радиусе и периферии:
$$c_1(r)=c_{1,ср}\left(\frac{r_{1,ср}}{r}\right)^{\cos^2\alpha_1},\qquad
T_1(r)=T_0^*-\frac{c_1(r)^2}{2000c_p},\qquad
T_{1ад}(r)=T_0^*-\frac{[c_1(r)/\varphi]^2}{2000c_p}.$$
По местному $p_1$ и принятому общему $p_2$ определяем $H_л(r)=c_pT_1(r)[1-(p_2/p_1(r))^{(k-1)/k}]$ и $\rho(r)=H_л(r)T_{1ад}(r)/(T_1(r)H_0)$. Для предварительной оценки РК сопоставляем одинаковые относительные высоты на входе и выходе и принимаем $c_{2a}$ и $p_2$ постоянными по высоте. Местные $u_1,u_2$ и $w_2$ считаем по формулам п. 5. Сравниваем корневую реактивность с нижним ориентиром 0,05.

**Источник.** Михальцев, Моляков, «Теория и проектирование газовой турбины», ч. 1, § 6.2, с. 58 и § 7.3, с. 76–77.

```python
radial_rows = []
for name, g in stages.items():
    a1 = math.radians(g['alpha1'])
    cp, k = g['cp'], g['k']
    for label, span in [('Корень',0.0),('Среднее',0.5),('Периферия',1.0)]:
        r1 = g['D_hub']/2 + span*g['l1']
        r2 = g['D_hub']/2 + span*g['l2']
        u1 = 2*math.pi*r1*g['n']/60
        u2 = 2*math.pi*r2*g['n']/60
        c1 = g['c1'] * (g['D1']/2/r1)**math.cos(a1)**2
        c1a, c1u = c1*math.sin(a1), c1*math.cos(a1)
        T1 = g['T0'] - c1**2/(2000*cp)
        T1ad = g['T0'] - (c1/phi)**2/(2000*cp)
        p1 = g['p0'] * (T1ad/g['T0'])**(k/(k-1))
        Hl = cp*T1*(1-(g['p2']/p1)**((k-1)/k))
        rho_local = Hl*T1ad/(T1*g['H0'])
        w1u = c1u-u1
        w1 = math.hypot(c1a,w1u)
        w2 = psi*math.sqrt(w1**2+2000*Hl+u2**2-u1**2)
        c2a = g['c2a']
        w2u = -math.sqrt(w2**2-c2a**2)
        T2 = T1+(u2**2-u1**2-w2**2+w1**2)/(2000*cp)
        radial_rows.append(dict(Турбина=name, Сечение=label, r1=r1, r2=r2,
                                rho=rho_local, c1a=c1a, c1u=c1u, w1u=w1u, c2a=c2a, w2u=w2u,
                                beta1=math.degrees(math.atan2(c1a,w1u)),
                                beta2=math.degrees(math.atan2(c2a,w2u)),
                                M1=c1/math.sqrt(1000*k*R_gas*T1),
                                Mw2=w2/math.sqrt(1000*k*R_gas*T2)))
radial = pd.DataFrame(radial_rows)
radial_display = radial[['Турбина','Сечение','r1','r2','rho','beta1','beta2','M1','Mw2']].copy()
radial_display[['r1','r2']] *= 1000
radial_display.columns = ['Турбина','Сечение','r1, мм','r2, мм','ρ','β1, град','β2, град','M1','Mw2']
display(radial_display)
display(radial_display.loc[radial_display['Сечение']=='Корень',['Турбина','ρ']]
        .assign(**{'ρ ≥ 0,05': lambda frame: frame['ρ'] >= 0.05}))
```

## 8. Вариантный расчёт для высоты лопаток от 20 мм

Цель — $l_{СА},l_{РК}\geq20$ мм для обеих турбин при общей втулке и $\sigma\leq300$ МПа. Значение 300 МПа задано пользователем для предварительного расчёта. На с. 57 методички предел около 300 МПа относится к особо напряжённым лопаткам; материал, температура и ресурс требуют отдельной проверки.

Исследуем $D_{вт}=30…80$ мм с шагом 5 мм, $\rho=0,20…0,50$ с шагом 0,05 и оценочную окружную скорость 250…450 м/с. Диапазоны реактивности и окружной скорости опираются на с. 58–59. Отношение $l/D$ вычисляется для каждого варианта, проверяются ориентир $1/3,5$ и предельное $1/3$ со с. 58. Оптимальное $u/c_0$ сравнивается с прежней приближённой аппроксимацией рис. 18.

На с. 56 скорость выхода 120–200 м/с указана для транспортных ГТД, 250–350 м/с — для авиационных в полёте, 80–120 м/с — для стационарных. Перебираем осевую скорость 100, 120, 140, 160, 180, 200, 250, 300, 350 м/с, а абсолютную выводим отдельно. При большой закрутке подмена абсолютной скорости осевой недопустима. Значение 100 м/с — расширение исследования за пределы транспортного диапазона.

Углы СА 10, 12, 15, 18, 21, 25° — исследовательская сетка, не установленный методичкой допустимый диапазон. Коэффициенты скорости 0,97/0,96 и зазор 0,5 мм сохранены. Для каждого варианта подбирается работа по балансу вала. Параметры выхода ТК передаются на вход СТ. Для СТ проверяются до трёх лучших по давлению вариантов ТК на каждый диаметр втулки. Такой отбор сокращает перебор и не доказывает глобальный оптимум.

Сначала проверяем размеры, напряжение и корневую реактивность. Затем отдельно оцениваем давление, скорость, числа Маха и углы. Критерии согласования ±2% по давлениям, ±5% по скорости СТ и углы выхода 80–100° приняты для сравнения вариантов и не являются цитатой из методички. Отсутствие подходящих строк означает отсутствие решения на проверенной сетке.

```python
from scipy.optimize import brentq
import itertools
saved_geometry = geometry.copy()
saved_hubs = hub_diameter.copy()
saved_n = n_selected.copy()
saved_angles = alpha1_selected.copy()
def calculate_variant(name, H0, T0, p0):
    rho = geometry.loc[name, 'ρ предварительная']
    n = n_selected[name]
    D_hub = hub_diameter.loc[name]
    alpha1 = math.radians(alpha1_selected[name])
    cp = cp_i(T0, alpha_design)
    for _ in range(20):
        T_mean = T0 - H0 * phi**2 / (2 * cp)
        cp = cp_i(T_mean, alpha_design)
    k = cp / (cp - R_gas)

    # Сопловой аппарат
    Hc = H0 * (1 - rho)
    c1 = phi * math.sqrt(2000 * Hc)
    T1 = T0 - phi**2 * Hc / cp
    T1_ad = T0 - Hc / cp
    p1 = p0 * (T1_ad / T0)**(k / (k - 1))
    v1 = R_gas * T1 / p1
    c1a = c1 * math.sin(alpha1)
    c1u = c1 * math.cos(alpha1)
    A1 = G_gas_design * v1 / c1a
    l1 = (math.sqrt(D_hub**2 + 4 * A1 / math.pi) - D_hub) / 2
    D1 = D_hub + l1
    u1 = math.pi * D1 * n / 60
    w1 = math.hypot(c1a, c1u - u1)

    # Рабочее колесо
    Hl = rho * H0 * T1 / T1_ad
    p2 = p1 * (1 - Hl / (cp * T1))**(k / (k - 1))
    c2a = ca_selected
    l2 = geometry.loc[name, 'l рабочей лопатки, мм'] / 1000
    for _ in range(60):
        D2 = D_hub + l2
        u2 = math.pi * D2 * n / 60
        w2 = psi * math.sqrt(w1**2 + 2000 * Hl + u2**2 - u1**2)
        T2 = T1 + (u2**2 - u1**2 - w2**2 + w1**2) / (2000 * cp)
        v2 = R_gas * T2 / p2
        A2 = G_gas_design * v2 / c2a
        l_next = (math.sqrt(D_hub**2 + 4 * A2 / math.pi) - D_hub) / 2
        if abs(l_next - l2) < 1e-12:
            break
        l2 = l_next
    else:
        raise ValueError(f'{name}: площадь за РК не сошлась')
    if c2a >= w2:
        raise ValueError(f'{name}: осевая скорость за РК выше относительной')
    w2u = -math.sqrt(w2**2 - c2a**2)
    c2u = u2 + w2u
    c2 = math.hypot(c2a, c2u)

    # Удельные потери в зазоре и на трение диска при полной подаче
    Lu = (u1 * c1u - u2 * c2u) / 1000
    hc = (1 / phi**2 - 1) * c1**2 / 2000
    hl = (1 / psi**2 - 1) * w2**2 / 2000
    hv = c2**2 / 2000
    hz = 1.37 * (1 + 1.6 * rho) * (1 + l2 / D2) * (radial_clearance / l2) * Lu
    density_mean = (1 / v1 + 1 / v2) / 2
    friction_power = 1.07 * D2**2 * (u2 / 100)**3 * density_mean
    h_friction = friction_power / G_gas_design
    L = Lu - hz - h_friction
    T_exit = T2 + (hz + h_friction) / cp
    T_exit_total = T_exit + hv / cp
    p_exit_total = p2 * (T_exit_total / T_exit)**(k / (k - 1))
    H_total = cp * T0 * (1 - (p_exit_total / p0)**((k - 1) / k))

    return dict(H0=H0, T0=T0, p0=p0, cp=cp, k=k, rho=rho, n=n,
                D_hub=D_hub, D1=D1, D2=D2, l1=l1, l2=l2, A1=A1, A2=A2,
                T1=T1, T2=T2, p1=p1, p2=p2, c1=c1, c1a=c1a, c1u=c1u,
                c2=c2, c2a=c2a, c2u=c2u, w1=w1, w2=w2, u1=u1, u2=u2,
                alpha1=alpha1_selected[name], alpha2=math.degrees(math.atan2(c2a,c2u)),
                beta1=math.degrees(math.atan2(c1a,c1u-u1)),
                beta2=math.degrees(math.atan2(c2a,w2u)),
                Lu=Lu, L=L, hc=hc, hl=hl, hv=hv, hz=hz, h_friction=h_friction,
                T2_total=T_exit_total, p2_total=p_exit_total, eta=L/H_total,
                eta_power=L/H0, sigma=kn*n**2*A2,
                M1=c1/math.sqrt(1000*k*R_gas*T1),
                M2w=w2/math.sqrt(1000*k*R_gas*T2))
```

```python
records=[]

for hub,ca,rho,u,angle in itertools.product(np.arange(.03,.081,.005),[100.,120.,140.,160.,180.,200.,250.,300.,350.],[.2,.25,.3,.35,.4,.45,.5],[250.,300.,350.,400.,450.],[10.,12.,15.,18.,21.,25.]):
    name='ТК'
    row=stage_inputs.loc[name]
    Hstart=preliminary.loc[name,'H0 оценка, кДж/кг']
    A=preliminary.loc[name,'A2 оценка, см²']/1e4*c2a_est[name]/ca
    l=(math.sqrt(hub*hub+4*A/math.pi)-hub)/2
    D=hub+l
    n_selected[name]=60*u/(math.pi*D)
    hub_diameter.loc[name]=hub
    geometry.loc[name,'ρ предварительная']=rho
    alpha1_selected[name]=angle
    ca_selected=ca
    try:
        H=brentq(lambda H:calculate_variant(name,H,row['T0*, К'],row['p0*, кПа'])['L']-row['L треб., кДж/кг'],.7*Hstart,1.6*Hstart,xtol=1e-5)
        g=calculate_variant(name,H,row['T0*, К'],row['p0*, кПа'])
        rec={k:float(g[k]) for k in ['H0','D_hub','l1','l2','D1','D2','rho','n','u2','c2','c2a','alpha1','alpha2','sigma','p2_total','T2_total','eta','M1','M2w']}
        rec['min_height']=min(g['l1'],g['l2'])
        rec['lD']=g['l2']/g['D2']
        rec['pressure_error']=g['p2_total']/design_cycle['p4']-1
        records.append(rec)
    except (ValueError,OverflowError): pass
df=pd.DataFrame(records)
ok=df[(df.min_height>=.020)&(df.sigma<=300)&(df.lD<=1/3)]
```

```python
def root_reaction(g):
    cp,k=g['cp'],g['k']
    c1=g['c1']*(g['D1']/g['D_hub'])**math.cos(math.radians(g['alpha1']))**2
    t=g['T0']-c1*c1/(2000*cp)
    ta=g['T0']-(c1/phi)**2/(2000*cp)
    if min(t,ta)<=0:return -1.
    p=g['p0']*(ta/g['T0'])**(k/(k-1))
    return cp*ta*(1-(g['p2']/p)**((k-1)/k))/g['H0']
roots=[]
for _,r in df.iterrows():
    name='ТК'
    n_selected[name]=r.n;hub_diameter.loc[name]=r.D_hub
    geometry.loc[name,'ρ предварительная']=r.rho;alpha1_selected[name]=r.alpha1;ca_selected=r.c2a
    g=calculate_variant(name,r.H0,stage_inputs.loc[name,'T0*, К'],stage_inputs.loc[name,'p0*, кПа'])
    roots.append(root_reaction(g))
df['rho_root']=roots
eligible=df[(df.min_height>=.020)&(df.sigma<=300)&(df.lD<=1/3)&(df.rho_root>=.05)]
pairs=[]
# Use up to three best TK candidates per common hub.
shortlist=eligible.sort_values('pressure_error',key=abs).groupby('D_hub').head(3)
for _,tk in shortlist.iterrows():
 for ca,rho,u,angle in itertools.product([100.,120.,160.,200.,250.,300.,350.],[.2,.3,.4,.5],[250.,300.,350.,400.,450.],[10.,12.,15.,18.,21.,25.,30.]):
    name='СТ';hub=tk.D_hub
    row=stage_inputs.loc[name];Hstart=preliminary.loc[name,'H0 оценка, кДж/кг']
    A=preliminary.loc[name,'A2 оценка, см²']/1e4*c2a_est[name]/ca
    l=(math.sqrt(hub*hub+4*A/math.pi)-hub)/2
    n_selected[name]=60*u/(math.pi*(hub+l))
    hub_diameter.loc[name]=hub
    geometry.loc[name,'ρ предварительная']=rho
    alpha1_selected[name]=angle
    ca_selected=ca
    try:
        H=brentq(lambda H:calculate_variant(name,H,tk.T2_total,tk.p2_total)['L']-row['L треб., кДж/кг'],.7*Hstart,1.6*Hstart,xtol=1e-5)
        g=calculate_variant(name,H,tk.T2_total,tk.p2_total)
        if min(g['l1'],g['l2'])<.020 or g['sigma']>300 or g['l2']/g['D2']>1/3: continue
        rec={'tk_'+k:float(v) for k,v in tk.items()}
        rec.update({'st_'+k:float(g[k]) for k in ['H0','D_hub','l1','l2','D1','D2','rho','n','u2','c2','c2a','alpha1','alpha2','sigma','p2_total','T2_total','eta','M1','M2w']})
        rec['st_rho_root']=root_reaction(g)
        rec['score']=abs(tk.pressure_error)+abs(g['p2_total']/design_cycle['p5_total']-1)+abs(g['c2']/c_pt_exit-1)
        pairs.append(rec)
    except (ValueError,OverflowError):pass
pair_df=pd.DataFrame(pairs)
```
